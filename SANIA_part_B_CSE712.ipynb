```python
import pandas as pd
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from sklearn.preprocessing import LabelEncoder

# Download necessary NLTK resources
nltk.download('punkt')        # Tokenizer models
nltk.download('stopwords')    # Stopwords list
nltk.download('wordnet')      # Lemmatizer

# Stopwords and lemmatizer initialization
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def preprocess_text(text):
    # Convert text to lowercase
    text = text.lower()
    # Tokenize text and remove punctuation
    words = nltk.word_tokenize(text)
    # Remove stopwords and lemmatize
    filtered_words = [lemmatizer.lemmatize(word) for word in words if word.isalnum() and not word in stop_words]
    return " ".join(filtered_words)

# Apply preprocessing to each text entry
df_cleaned['text'] = df_cleaned['text'].apply(preprocess_text)
```

```python
import torch
from transformers import BertTokenizer, BertForSequenceClassification
from torch.utils.data import Dataset, DataLoader
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

# Load tokenizer
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# Initialize BERT model for sequence classification with 4 classes
model = BertForSequenceClassification.from_pretrained('bert-base-uncased', num_labels=4)

# Encode the labels into integers
label_encoder = LabelEncoder()
df_cleaned['label_category'] = label_encoder.fit_transform(df_cleaned['label_category'])

# Custom dataset
class TextDataset(Dataset):
    def __init__(self, texts, labels, tokenizer, max_len=128):
        self.texts = texts
        self.labels = labels
        self.tokenizer = tokenizer
        self.max_len = max_len

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, index):
        text = self.texts[index]
        labels = self.labels[index]

        encoding = tokenizer.encode_plus(
            text,
            add_special_tokens=True,
            max_length=self.max_len,
            return_token_type_ids=False,
            padding='max_length',
            truncation=True,
            return_attention_mask=True,
            return_tensors='pt',
        )

        return {
            'input_ids': encoding['input_ids'].flatten(),
            'attention_mask': encoding['attention_mask'].flatten(),
            'labels': torch.tensor(labels, dtype=torch.long)
        }

# Prepare data loaders
def create_data_loader(df, tokenizer, max_len, batch_size):
    ds = TextDataset(
        texts=df.text.to_numpy(),
        labels=df.label_category.to_numpy(),
        tokenizer=tokenizer,
        max_len=max_len
    )

    return DataLoader(ds, batch_size=batch_size, num_workers=2)

# Split data
train_df, test_df = train_test_split(df_cleaned, test_size=0.1, random_state=42)
train_loader = create_data_loader(train_df, tokenizer, max_len=128, batch_size=16)
test_loader = create_data_loader(test_df, tokenizer, max_len=128, batch_size=16)

# Training function (simplified)
def train_epoch(model, data_loader, loss_fn, optimizer, device, scheduler, n_examples):
    model = model.train()
    losses = []
    correct_predictions = 0

    for d in data_loader:
        input_ids = d["input_ids"].to(device)
        attention_mask = d["attention_mask"].to(device)
        labels = d["labels"].to(device)

        outputs = model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            labels=labels
        )

        _, preds = torch.max(outputs.logits, dim=1)
        loss = outputs.loss

        correct_predictions += torch.sum(preds == labels)
        losses.append(loss.item())

        loss.backward()
        optimizer.step()
        optimizer.zero_grad()
        scheduler.step()

    return correct_predictions.double() / n_examples, np.mean(losses)

# Example usage
# Define optimizer, loss function, scheduler, and device (similar to what was shown above in training and evaluation)
```

```python
def evaluate(model, data_loader, device):
    model.eval()  # Set model to evaluation mode

    total_predictions = torch.tensor([]).to(device)
    total_true_labels = torch.tensor([]).to(device)

    with torch.no_grad():
        for batch in data_loader:
            input_ids = batch['input_ids'].to(device)
            attention_mask = batch['attention_mask'].to(device)
            labels = batch['labels'].to(device)

            outputs = model(input_ids=input_ids, attention_mask=attention_mask, labels=labels)
            _, preds = torch.max(outputs.logits, dim=1)

            total_predictions = torch.cat((total_predictions, preds), dim=0)
            total_true_labels = torch.cat((total_true_labels, labels), dim=0)

    correct_predictions = torch.sum(total_predictions == total_true_labels)
    accuracy = correct_predictions.double() / total_true_labels.shape[0]
    return accuracy.item()
```

```python
from transformers import AdamW, get_scheduler

optimizer = AdamW(model.parameters(), lr=5e-5)
num_epochs = 3
num_training_steps = num_epochs * len(train_loader)
scheduler = get_scheduler(
    "linear",
    optimizer=optimizer,
    num_warmup_steps=0,
    num_training_steps=num_training_steps
)
```

```python
from sklearn.metrics import f1_score, classification_report
import torch
from torch.optim.lr_scheduler import ReduceLROnPlateau

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)

# To track loss and performance history
losses = []
best_accuracy = 0.0

# Scheduler that reduces learning rate when a metric has stopped improving
scheduler = ReduceLROnPlateau(optimizer, 'min', patience=2, factor=0.1, verbose=True)

for epoch in range(num_epochs):
    model.train()
    all_predictions = []
    all_true_labels = []

    total_loss = 0
    num_batches = 0

    for batch in train_loader:
        batch = {k: v.to(device) for k, v in batch.items()}
        outputs = model(**batch)

        loss = outputs.loss
        total_loss += loss.item()
        num_batches += 1

        loss.backward()
        optimizer.step()
        optimizer.zero_grad()

        # Convert logits to predictions
        _, predictions = torch.max(outputs.logits, dim=1)
        all_predictions.extend(predictions.cpu().numpy())
        all_true_labels.extend(batch['labels'].cpu().numpy())

    # Calculate average loss for the epoch
    average_loss = total_loss / num_batches
    losses.append(average_loss)

    # Evaluate and calculate accuracy and F1 score
    accuracy = evaluate(model, test_loader, device)
    f1 = f1_score(all_true_labels, all_predictions, average='weighted')

    # Check for early stopping and learning rate reduction
    scheduler.step(average_loss)  # Adjust learning rate based on the validation loss
    if accuracy > best_accuracy:
        best_accuracy = accuracy
        torch.save(model.state_dict(), 'best_model_state.bin')  # Save the best model
        print(f"Saved Improved Model at Epoch {epoch + 1}: Accuracy = {accuracy:.4f}")
    else:
        print("No improvement in accuracy. Best Accuracy: {:.4f}".format(best_accuracy))

    print(f"Epoch {epoch + 1}/{num_epochs} | Loss: {average_loss:.4f} | Accuracy: {accuracy:.4f} | F1 Score: {f1:.4f}")

# Load the best model (optional, if further validation/inference is needed post-training)
model.load_state_dict(torch.load('best_model_state.bin'))

# Optionally, print the classification report for the last epoch
print("\nDetailed Classification Report:")
print(classification_report(all_true_labels, all_predictions))
```
